# FIN·LAB: consultar la API desde Google Colab

Este cuaderno consume la API educativa publicada en Vercel. Ya apunta a la API del curso; si publicás tu propio despliegue, cambiá la URL de `API_URL`. Los valores son una muestra fija, no cotizaciones en tiempo real.

```python
import requests
import pandas as pd

API_URL = "https://postgrado-ia.vercel.app/api/market/summary"
response = requests.get(API_URL, timeout=15)
response.raise_for_status()
payload = response.json()
assert payload["ok"], payload
print("Sesión:", payload["data"]["session"])
print("Fuente:", payload["meta"]["source"])
pd.DataFrame(payload["data"]["tickers"])[["symbol", "last", "change_pct", "rsi14", "signal"]]
```
